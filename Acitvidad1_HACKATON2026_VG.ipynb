## Librerías

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
```

## Conexión a la base de datos OBIS

La API de OBIS recibe un área en formato **WKT** (*Well Known Text*). Definimos un polígono aproximado que cubre las aguas costeras del **Pacífico de Guatemala**

```python
url_obis = "https://api.obis.org/v3/occurrence"

pacifico = "POLYGON((-92.30 13.00, -90.05 13.00, -90.05 14.60, -92.30 14.60, -92.30 13.00))"
```

```python
#Los parámetros de la consulta: el área, cuántos registros queremos (`size`) y qué columnas (`fields`).

parametros = {
    "geometry": pacifico,
    "size": 5000,
    "fields": "id,scientificName,phylum,class,order,family,genus,species,"
              "decimalLatitude,decimalLongitude,eventDate,date_year,depth,basisOfRecord",
}

# Hacemos la consulta
respuesta = requests.get(url_obis, params=parametros, timeout=120)
respuesta.status_code
```

```python
# Descargamos los datos

datos = respuesta.json()
registros = datos["results"]
len(registros)
```

## Guardar el archivo `.csv`

```python
datos_obis = pd.DataFrame(registros)
datos_obis.to_csv("registros_marinos_pacifico_guatemala.csv", index=False)
```

## Leer el archivo `.csv`
